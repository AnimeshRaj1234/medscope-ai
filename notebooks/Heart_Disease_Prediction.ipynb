#### INSTALL LIB

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

#### IMPORT DATASET

```python
data = pd.read_csv('../datasets/heart.csv')
```

#### DATA PREPROCESS

```python
X = data.drop("target", axis=1)
y = data["target"]
```

#### TRAIN-TEST-SPLIT

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.3,random_state=42)
```

##### FEATURE SCALING

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

#### MODEL TRAINING 

```python
# Logistic Regression
lr = LogisticRegression(max_iter=1000)

lr.fit(X_train,y_train)
```

```python
# Random Forest
rf = RandomForestClassifier(n_estimators=100,random_state=42)

rf.fit(X_train,y_train)
```

#### PREDICTIONS

```python
lr_pred = lr.predict(X_test);

rf_pred = rf.predict(X_test);
```

#### EVALUATION METRICS

```python
# ACCURACY

print("ACCURACY SCORE LR : \n",accuracy_score(y_test,lr_pred)) # Logistic Regression

print ("\n")

print ("ACCURACY SCORE RF : \n",accuracy_score(y_test,rf_pred)) # Random Forest

```

```python
# CLASSIFICATION REPORT

print ("CLASSIFICATION REPORT OF LR : \n", classification_report(y_test,lr_pred)) # logistic regrssion

print ("\n")

print ("CLASSIFICATION REPORT OF RF : \n",classification_report(y_test,rf_pred)) # random forest
```

```python
# CONFUSION MATRIX

print ("CONFUSION MATRIX OF LR : \n") # logistic regrssion

cm_lr = confusion_matrix(y_test, lr_pred)

sns.heatmap(cm_lr, annot=True, fmt='d')
plt.show()

print("\n")

print ("CONFUSION MATRIX OF RF : \n") # random forest

cm_rf = confusion_matrix(y_test, rf_pred)

sns.heatmap(cm_rf, annot=True, fmt='d')
plt.show()
```

#### FEATURE IMPORTANCE FOR RANDOM FOREST

```python
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': rf.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

top_feature = feature_importance.head(5) # top five features

# Plot
plt.figure(figsize=(10, 6))
sns.barplot(
    x='Feature',
    y='Importance',
    data=top_feature
)

plt.title('Feature Importance - Random Forest')
plt.xlabel('Features')
plt.ylabel('Importance Score')
plt.tight_layout()
plt.show()
```

## Model Comparison

| Model | Accuracy |
|---------|---------|
| Logistic Regression | 79.51% |
| Random Forest | 98.54% |

**Observation:** Random Forest significantly outperformed Logistic Regression and was selected as the final model.

## Conclusion

A Heart Disease Prediction model was successfully developed using machine learning techniques. After data exploration, preprocessing, and model training, both Logistic Regression and Random Forest classifiers were evaluated. The Random Forest model achieved the best performance with an accuracy of 98.54%.

Feature importance analysis revealed that Chest Pain Type (cp), Number of Major Vessels (ca), Maximum Heart Rate (thalach), ST Depression (oldpeak), and Thalassemia (thal) were the most influential factors in predicting heart disease.

The results demonstrate the effectiveness of machine learning in assisting early heart disease detection and supporting healthcare decision-making.
